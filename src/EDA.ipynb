# PERFORMING EDA ON AMOZON SALE DATASET

```python
# import librariear
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Load dataset (amazon sales)

```python
df=pd.read_csv('../data/amazon.csv')
```

Performing EDA


```python
df.head(5)
df.describe()
print('The shape of the dataset is:', df.shape)
print('The columns in the dataset are:', df.columns)
print('The data types of the columns are:', df.dtypes)
```

```python
df.isnull().sum()
```

```python
#convert discount_price,original_price, and discount_percentage  to numeric values
#convert rating to float    
df['discount_price'] = df['discount_price'].str.replace('₹', '').str.replace(',', '').astype(float)
df['original_price'] = df['original_price'].str.replace('₹', '').str.replace(',', '').astype(float)
df['discount_percentage'] = df['discount_percentage'].str.replace('%', '').astype(float)
df['rating'] = df['rating'].astype(float)
```

```python
df.head(5)
```

```python
# Visualize the distribution of ratings
plt.figure(figsize=(10, 6))
sns.histplot(df['rating'], bins=20, kde=True)
plt.title('Distribution of Ratings')
plt.xlabel('Rating')
plt.ylabel('Frequency')
plt.show()
```
